```python

start = time.time()
sim.stop_server()
# start the visualization server: in your browser go to http://localhost:1234
sim.start_server(port=1276)
sim.move_to_hel()
#sim.integrator = "trace"
#sim.integrator = "whfast"
sim.integrator = "leapfrog"
sim.dt =0.14#3.2# 0.005
sim.softening = 0.01 #(Roche limit?)

#sim.ri_whfast.corrector = 17
#sim.ri_whfast.safe_mode = 0
#sim.ri_whfast.coordinates = "jacobi"
#sim.ri_whfast.recalculate_coordinates_this_timestep = 1

times = np.linspace(0, 1000, 1000)



#sim.ri_trace.r_crit_hill = 1#2
#sim.ri_trace.corrector = 17
#sim.ri_trace.safe_mode = 0

#sim.N_active = 6#N_planets + 1
sim.collision = "none"
#sim.exit_min_distance = 0.006
#sim.exit_max_distance = 1500
sim.collision = "direct"
sim.collision_resolve = "merge"
sim.collision_radius = 0.0001
#sim.track_particle = 'Neptune'
boxsize = 4200
sim.configure_box(boxsize)
sim.boundary = "open"
#sim.move_to_hel()

au1=[]
au01=[]
au001=[]

au_a001=[]
au_a01=[]

c=0
d10=0
sim.move_to_hel()
#for step in range(10000000):
while sim.t<1.49e9:
    sim.move_to_hel()
    sim.integrate(sim.t + 400*sim.dt)
    sim.move_to_hel()
    to_remove=[]
    for i, p in enumerate(sim.particles[1:]):
        sim.move_to_hel()
        r = np.sqrt(p.x**2 + p.y**2+p.z**2)
        if r > max_distance:
            to_remove.append(i+1)
        elif r < min_distance:
            sim.move_to_hel()
            to_remove.append(i+1)
            au001.append([sim.t,p.a,p.e])
            print('Roche Limit')

        elif r < 0.1:
            sim.move_to_hel()
            au01.append([sim.t,p.a,p.e])
            print('0.1 au')
        elif r < 1:
            sim.move_to_hel()
            au1.append([sim.t,p.a,p.e])
            
            print('1 au')
        elif r<=3:
            sim.move_to_hel()
            au1.append([sim.t,p.a,p.e])
           # print('10 au')
            d10=1
        
        if np.abs(p.a) < min_distance:
            sim.move_to_hel()
            au_a001.append([sim.t,p.a,p.e,p.hash.value,p.d])
            print('Roche a')
        elif np.abs(p.a) < 0.1:
            sim.move_to_hel()
            au_a01.append([sim.t,p.a,p.e,p.hash.value,p.d])
            print('a<0.1')

    for idx in reversed(to_remove):
        sim.remove(idx)
        sim.move_to_hel()
       # print('N=',sim.N)
    if c<10000 and d10==0 :
        c+=1
    elif c>=10000 or d10==1:
        sim.move_to_hel()
        a_sJupiter.append(sim.particles['Jupiter'].a)
        a_sSaturn.append(sim.particles['Saturn'].a)
        #a_sUranus.append(sim.particles['Uranus'].a)
        #a_sNeptune.append(sim.particles['Neptune'].a)
        a_sMS.append(sim.particles['MS'].a)
    
        e_sJupiter.append(sim.particles['Jupiter'].e)
        e_sSaturn.append(sim.particles['Saturn'].e)
        #e_sUranus.append(sim.particles['Uranus'].e)
        #e_sNeptune.append(sim.particles['Neptune'].e)
        e_sMS.append(sim.particles['MS'].e)

        xyz_sJupiter.append([sim.particles['Jupiter'].x,sim.particles['Jupiter'].y,sim.particles['Jupiter'].z])
        xyz_sSaturn.append([sim.particles['Saturn'].x,sim.particles['Saturn'].y,sim.particles['Saturn'].z])
        #xyz_sUranus.append([sim.particles['Uranus'].x,sim.particles['Uranus'].y,sim.particles['Uranus'].z])
        #xyz_sNeptune.append([sim.particles['Neptune'].x,sim.particles['Neptune'].y,sim.particles['Neptune'].z])
        xyz_sMS.append([sim.particles['MS'].x,sim.particles['MS'].y,sim.particles['MS'].z])

        vxyz_sJupiter.append([sim.particles['Jupiter'].vx,sim.particles['Jupiter'].vy,sim.particles['Jupiter'].vz])
        vxyz_sSaturn.append([sim.particles['Saturn'].vx,sim.particles['Saturn'].vy,sim.particles['Saturn'].vz])
       # vxyz_sUranus.append([sim.particles['Uranus'].vx,sim.particles['Uranus'].vy,sim.particles['Uranus'].vz])
        #vxyz_sNeptune.append([sim.particles['Neptune'].vx,sim.particles['Neptune'].vy,sim.particles['Neptune'].vz])
        vxyz_sMS.append([sim.particles['MS'].vx,sim.particles['MS'].vy,sim.particles['MS'].vz])
        
        c=0
        d10=0

        ts.append(sim.t)
        sim.move_to_hel()
    sim.move_to_hel()
#sim.move_to_com()
sim.move_to_hel() 
#sim.integrate(1000)

#try:
#sim.integrate(1000)
#except rebound.Escape as e:
 #   print("Particle escaped")
  #  for i in reversed(range(1, sim.N)):
   #     if sim.particles[i].d>=1500:
    #       j=sim.particles[i].hash.value
     #      sim.remove(hash=int(j)) 
#except rebound.Encounter as e:
 #   print("Close encounter detected")
        
sim.stop_server(port=1276)

print(start-time.time())
```

```python
def current_a_e(x,y,z,vx,vy,vz,m):
    G=4*(np.pi**2)
    rvec=np.array([x,y,z])
    vvec=np.array([vx,vy,vz])
    v=np.sqrt((vx**2)+(vy**2)+(vz**2))
    r=np.sqrt((x**2)+(y**2)+(z**2))
    hx=y*vz-z*vy
    hy=z*vx-x*vz
    hz=x*vy-y*vx
    ex=(1/(m*G))*(vy*hz-vz*hy)-x/r
    ey=(1/(m*G))*(vz*hx-vx*hz)-y/r
    ez=(1/(m*G))*(vx*hy-vy*hx)-z/r
 
    e=np.sqrt(ex**2+ey**2+ez**2)

    a=((2/r)-(v**2/(G*m)))**-1

    n=np.array([-hy,hx,0])
    evec=np.array([ex,ey,ez])
    hvec=np.array([hx,hy,hz])
    
    if ez<0:
        om=2*np.pi-np.arccos(n@evec/np.sqrt(sum(n**2)*sum(evec**2)))
    else:
        om=np.arccos(n@evec/np.sqrt(sum(n**2)*sum(evec**2)))

    if rvec@vvec<0:
        f=2*np.pi-np.arccos(rvec@evec/(r*e))
    else:
        f=np.arccos(rvec@evec/(r*e))

    if hx<0:
        Om=2*np.pi-np.arccos(-hy/np.sqrt(sum(n**2)))
    else:
        Om=np.arccos(-hy/np.sqrt(sum(n**2)))

    i=np.arccos(hz/np.sqrt(sum(hvec**2)))

    return a,e,om,f,i,Om
        
```
